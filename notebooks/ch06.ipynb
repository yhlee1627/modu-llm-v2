# 6장. LLM은 어떻게 배우고 말할까요?

문장마다 손실(모델이 놀란 정도)을 재고, 학습할수록 손실이 어떻게 바뀌는지 봅니다. 온도와 top-k를 바꿔 다음 글자 고르기가 어떻게 달라지는지도 확인합니다.

- GPU가 없어도 됩니다.
- 따옴표 안의 글과 숫자만 바꿔 보세요.

데이터: [Tiny-Ko-Stories](https://huggingface.co/datasets/psymon/Tiny-Ko-Stories) (psymon, CC BY 4.0) · 모델: [JosephLee/modu-llm](https://huggingface.co/JosephLee/modu-llm) (CC BY 4.0)

## 실습 6-1 이야기 LLM 불러오기

```python
%pip install -q git+https://github.com/yhlee1627/modu-llm-v2
from modu_llm import load_story_llm, compare_stages
llm = load_story_llm()
```

## 실습 6-2 글자마다 손실 보기

마지막 줄 '지'의 확률과 손실을 계산기로 확인해 보세요. 손실 = −ln(확률)입니다.

```python
llm.losses("어느 날, 지")
```

## 실습 6-3 문장마다 손실 비교하기

**먼저 예상해 보세요.** 네 문장 가운데 손실이 가장 작은 문장과 가장 큰 문장은 무엇일까요?

```python
llm.compare_losses(["토끼가 숲에 가요.", "숲에 토끼가 가요.", "토끼가 숲에 갑요.", "요가 숲토 에끼."])
```

## 실습 6-4 어느 글자에서 놀랐을까?

**먼저 예상해 보세요.** "토끼가 숲에 갑요."에서 손실이 가장 큰 글자는 무엇일까요?

```python
llm.losses("토끼가 숲에 갑요.")
```

## 실습 6-5 학습할수록 손실은?

**먼저 예상해 보세요.** 학습이 진행될수록 두 문장의 손실은 각각 어떻게 될까요? 모델 다섯 개를 차례로 내려받으므로 시간이 조금 걸립니다.

```python
compare_stages(["어느 날, 토끼는 숲으로 놀러 갔어요.", "요가 숲토 에끼."])
```

## 실습 6-6 온도를 바꾼 확률

**먼저 예상해 보세요.** 온도를 0.5로 낮추면 1위 '지'의 확률은 커질까요, 작아질까요? 2.0으로 높이면요?

```python
llm.next_chars("어느 날, ", temperature=0.5)
llm.next_chars("어느 날, ", temperature=2.0)
```

## 실습 6-7 온도를 낮춰 이야기 쓰기

온도 0.3으로 같은 시작 글에서 두 번 씁니다.

```python
llm.write("어느 날, ", length=60, temperature=0.3, n=2)
```

## 실습 6-8 온도를 높이고 top-k로 막기

온도 1.5로 쓴 뒤, 같은 온도에 top-k 5를 더해 씁니다.

```python
llm.write("어느 날, ", length=60, temperature=1.5)
llm.write("어느 날, ", length=60, temperature=1.5, top_k=5)
```
