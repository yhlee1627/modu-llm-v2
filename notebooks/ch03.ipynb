## 실습 3-1 이야기 LLM 불러오기

```python
%pip install -q git+https://github.com/yhlee1627/modu-llm-v2
from modu_llm import load_story_llm
llm = load_story_llm()
```

## 실습 3-2 글자를 번호로 바꾸기

**먼저 예상해 보세요.** "토끼가 숲에 가요."에는 '가'가 두 번 나옵니다. 두 '가'의 번호는 같을까요? 띄어쓰기에도 번호가 있을까요?

```python
llm.to_ids("토끼가 숲에 가요.")
```

## 실습 3-3 번호를 글자로 바꾸기

**먼저 예상해 보세요.** 실습 3-2에서 '가'는 102번, '요'는 1579번이었습니다. 두 번호를 글자로 바꾸면 무엇이 나올까요?

```python
llm.to_text([102, 1579])
```

## 실습 3-4 사전에 없는 글자

**먼저 예상해 보세요.** '똠'과 웃는 얼굴 그림 글자 😀는 글자 사전에 있을까요?

```python
llm.to_ids("똠얌꿍 좋아 😀")
```

## 실습 3-5 숫자 카드 보기

**먼저 예상해 보세요.** 3번째 '가'와 8번째 '가'의 글자 카드는 같을까요? 두 카드에 자리 카드를 더한 합친 카드는요?

```python
llm.cards("토끼가 숲에 가요.")
```

## 실습 3-6 실제 LLM과 토큰 비교하기

네이버가 공개한 한국어 LLM의 토크나이저를 내려받아, 같은 글을 이야기 LLM과 어떻게 다르게 나누는지 봅니다. 처음 실행하면 내려받는 데 시간이 조금 걸립니다.

```python
llm.compare_tokens("어느 날, 토끼는 숲으로 산책을 갔어요.", "naver-hyperclovax/HyperCLOVAX-SEED-Text-Instruct-0.5B")
```

## 실습 3-7 다른 LLM과도 비교하기

OpenAI가 공개한 LLM의 토크나이저로 같은 글을 나눠 봅니다. 따옴표 안의 글을 영어 문장으로 바꿔서도 해 보세요.
