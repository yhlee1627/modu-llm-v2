# 4장. 앞의 글자를 참고하는 어텐션

학습을 마친 이야기 LLM 안에서 어텐션이 만든 비율 표를 직접 봅니다. 뒤 글자가 가려져 있는지, 줄마다 비율의 합이 1인지 확인합니다.

- GPU가 없어도 됩니다.
- 따옴표 안의 글과 숫자만 바꿔 보세요.

데이터: [Tiny-Ko-Stories](https://huggingface.co/datasets/psymon/Tiny-Ko-Stories) (psymon, CC BY 4.0) · 모델: [JosephLee/modu-llm](https://huggingface.co/JosephLee/modu-llm) (CC BY 4.0)

## 실습 4-1 이야기 LLM 불러오기

```python
%pip install -q git+https://github.com/yhlee1627/modu-llm-v2
from modu_llm import load_story_llm
llm = load_story_llm()
```

## 실습 4-2 비율 표 보기

**먼저 예상해 보세요.** 비율 표의 오른쪽 위 칸들에는 어떤 값이 있을까요? 줄마다 비율을 모두 더하면 얼마일까요?

```python
llm.attention("토끼가 숲에 가요.", layer=1, head=2, total=True)
```

## 실습 4-3 한 글자의 계산 따라가기

3번째 글자 '가'가 앞 글자를 참고하는 계산을 단계마다 봅니다.

```python
llm.attention_steps("토끼가 숲에 가요.", layer=1, head=2, at=3)
```

## 실습 4-4 새 문장에서도 그럴까?

**먼저 예상해 보세요.** 층 1 헤드 2는 새 문장에서도 바로 앞 글자만 볼까요?

```python
llm.looks_at("어느 날, 토끼는 숲으로 갔어요.", layer=1, head=2)
```

## 실습 4-5 같은 층의 다른 헤드

**먼저 예상해 보세요.** 같은 층 1의 헤드 1도 바로 앞 글자를 볼까요? head 뒤의 숫자를 1~4로 바꿔 가며 보세요.

```python
llm.looks_at("어느 날, 토끼는 숲으로 갔어요.", layer=1, head=1)
```
