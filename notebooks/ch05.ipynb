## 실습 5-1 이야기 LLM 불러오기

```python
%pip install -q git+https://github.com/yhlee1627/modu-llm-v2
from modu_llm import load_story_llm
llm = load_story_llm()
```

## 실습 5-2 헤드 24개 한눈에 보기

**먼저 예상해 보세요.** 6층 × 4개, 헤드 24개가 모두 같은 곳을 볼까요?

```python
llm.heads_summary("어느 날, 토끼는 숲으로 놀러 갔어요. 토끼는 곰을 만났어요.")
```

## 실습 5-3 쉼표를 보는 헤드

층 3 헤드 2가 글자마다 가장 많이 참고하는 글자를 봅니다.

```python
llm.looks_at("어느 날, 토끼는 숲으로 놀러 갔어요.", layer=3, head=2)
```

## 실습 5-4 띄어쓰기를 보는 헤드

층 6 헤드 1이 글자마다 가장 많이 참고하는 글자를 봅니다.

```python
llm.looks_at("어느 날, 토끼는 숲으로 놀러 갔어요.", layer=6, head=1)
```

## 실습 5-5 층마다 다음 글자 맞혀 보기

**먼저 예상해 보세요.** "토끼는 숲으로 산" 다음 글자를 임베딩만 거친 뒤에도 맞힐 수 있을까요? 몇 층쯤에서 '책'이 1위가 될까요?

```python
llm.layer_guess("토끼는 숲으로 산")
```
