```python
%pip install -q git+https://github.com/yhlee1627/modu-llm-v2
from modu_llm import load_story_llm
llm = load_story_llm()
```

## 실습 5-2 헤드 24개 한눈에 보기

**먼저 예상해 보세요.** 블록 6층에 헤드가 4개씩, 모두 24개입니다. 헤드 24개가 모두 같은 곳을 볼까요?

```python
llm.heads_summary("어느 날, 토끼는 숲으로 놀러 갔어요. 토끼는 곰을 만났어요.")
```

## 실습 5-3 쉼표를 보는 헤드

**먼저 예상해 보세요.** 쉼표 뒤의 글자들은 쉼표를 볼 것 같습니다. 그렇다면 쉼표가 나오기 전의 '어', '느', '날'은 무엇을 볼까요?

```python
llm.looks_at("어느 날, 토끼는 숲으로 놀러 갔어요.", layer=3, head=2)
```

## 실습 5-4 띄어쓰기를 보는 헤드

**먼저 예상해 보세요.** 층 3 헤드 2가 쉼표를 보듯, 이 헤드도 글자마다 띄어쓰기를 또렷하게 볼까요?

```python
llm.looks_at("어느 날, 토끼는 숲으로 놀러 갔어요.", layer=6, head=1)
```

## 실습 5-5 층마다 다음 글자 맞혀 보기

**먼저 예상해 보세요.** "토끼는 숲으로 산" 다음 글자를 숫자 카드만 거친 뒤에도 맞힐 수 있을까요? 몇 층쯤에서 '책'이 1위가 될까요?

```python
llm.layer_guess("토끼는 숲으로 산")
```
