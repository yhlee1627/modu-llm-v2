## 실습 1-1 이야기 LLM 불러오기

패키지를 설치하고, 사전학습을 마친 이야기 LLM을 불러옵니다.

```python
%pip install -q git+https://github.com/yhlee1627/modu-llm-v2
from modu_llm import load_story_llm
llm = load_story_llm()
```

## 실습 1-2 다음 글자 후보 보기

"어느 날, " 다음 글자 후보와 확률을 직접 확인합니다.

```python
llm.next_chars("어느 날, ")
```

## 실습 1-3 다른 글로 후보 보기

**먼저 예상해 보세요.** "토끼는 숲으로 " 다음에는 어떤 글자가 올까요? 세 글자를 적어 본 뒤 실행합니다.

```python
llm.next_chars("토끼는 숲으로 ")
```

## 실습 1-4 한 글자 붙이고 다시 맞히기

**먼저 예상해 보세요.** 1위 후보 '산'을 붙이면 다음 글자는 무엇일까요? 1위의 확률은 커질까요, 작아질까요?

```python
llm.next_chars("토끼는 숲으로 산")
```

## 실습 1-5 이어서 이야기 쓰기

맞히기 → 고르기 → 붙이기를 100번 되풀이합니다.

```python
llm.write("어느 날, ")
```

## 실습 1-6 같은 시작, 다른 이야기

같은 "어느 날, "로 세 번 써 봅니다.

```python
llm.write("어느 날, ", length=60, n=3)
```

## 실습 1-7 다른 시작 글로 이야기 쓰기

따옴표 안의 시작 글과 length 뒤의 숫자를 바꿔 실행해 보세요. 예: `"옛날 옛적에 "`, `"비가 오는 날, "`

```python
llm.write("토끼는 숲으로 ", length=80)
```
