## 실습 1-3 다른 글로 후보 보기

**먼저 예상해 보세요.** "토끼는 숲으로 " 다음에는 어떤 글자가 올까요? 세 글자를 적어 본 뒤 실행합니다.

```python
llm.next_chars("토끼는 숲으로 ")
```

## 실습 1-4 한 글자 붙이고 다시 맞히기

**먼저 예상해 보세요.** 1위 후보 '산'을 붙이면 다음 글자는 무엇일까요? 1위의 확률은 커질까요, 작아질까요?

```python
llm.next_chars("토끼는 숲으로 산")
```
