### 실습 A-11 if 문

```python
s = "토끼가 숲에 가요"
spaces = 0
for ch in s:
    if ch == " ":
        spaces = spaces + 1
print("띄어쓰기:", spaces)

p = 0.118
if p > 0.5:
    print("거의 확실해요")
else:
    print("아직 망설여요")
```

## A.6 클래스

### 실습 A-14 붕어빵 틀

```python
class FishBread:
    def __init__(self, filling):
        self.filling = filling

    def describe(self):
        return self.filling + " 붕어빵"

bread1 = FishBread("팥")
bread2 = FishBread("슈크림")
print(bread1.describe())
print(bread2.filling)
```

### 실습 A-15 PyTorch 부품 만들기

```python
import torch
from torch import nn

class Times(nn.Module):
    def __init__(self, k):
        super().__init__()
        self.k = k

    def forward(self, x):
        return x * self.k

triple = Times(3)
print(triple(torch.tensor([1.0, 2.0, 3.0])))
```
