## 실습 2-1 세 모델 불러오기

학습 전(before), 사전학습 뒤(pretrain), 파인튜닝 뒤(sft)의 이야기 LLM을 불러옵니다.

```python
%pip install -q git+https://github.com/yhlee1627/modu-llm-v2
from modu_llm import load_story_llm
before = load_story_llm("before")
pre = load_story_llm("pretrain")
sft = load_story_llm("sft")
```

## 실습 2-2 학습 전 모델의 다음 글자 후보

**먼저 예상해 보세요.** 학습 전 모델에서 1위 후보의 확률은 얼마쯤일까요?

```python
before.next_chars("어느 날, ")
```

## 실습 2-3 학습 전 모델로 이어 쓰기

```python
before.write("어느 날, ", length=60)
```

## 실습 2-4 사전학습 도중의 모델

**먼저 예상해 보세요.** 250걸음, 1,000걸음, 3,000걸음 학습한 모델의 글은 어떻게 다를까요?

```python
load_story_llm("step250").write("어느 날, ", length=80)
load_story_llm("step1000").write("어느 날, ", length=80)
load_story_llm("step3000").write("어느 날, ", length=80)
```

## 실습 2-5 사전학습 모델에 요청하기

**먼저 예상해 보세요.** 사전학습만 한 모델은 요청대로 토끼 이야기를 쓸까요?

```python
pre.ask("토끼가 나오는 짧은 이야기를 써 줘", length=150)
```

## 실습 2-6 파인튜닝 모델에 같은 요청하기

```python
sft.ask("토끼가 나오는 짧은 이야기를 써 줘", length=150)
```

## 실습 2-7 이야기가 아닌 질문하기

**먼저 예상해 보세요.** 파인튜닝 모델은 "서울"이라고 답할까요?

```python
sft.ask("대한민국의 수도는 어디야?", length=150)
```

## 실습 2-8 배우지 않은 말

**먼저 예상해 보세요.** "오늘부터 진짜 다이어" 다음 후보에 '트'가 있을까요?

```python
pre.next_chars("오늘부터 진짜 다이어")
```
