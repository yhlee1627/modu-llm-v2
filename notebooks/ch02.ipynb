## 실습 2-1 네 모델 불러오기

```python
%pip install -q git+https://github.com/yhlee1627/modu-llm-v2
from modu_llm import load_story_llm
before = load_story_llm("before")
pre = load_story_llm("pretrain")
sft = load_story_llm("sft")
fb = load_story_llm("feedback")
```

## 실습 2-2 학습 전 모델의 다음 글자 후보

**먼저 예상해 보세요.** 학습 전 모델에게 "어느 날, " 다음 글자를 물으면, 1위 후보의 확률은 얼마쯤일까요?

```python
before.next_chars("어느 날, ")
```

## 실습 2-3 학습 전 모델로 이어 쓰기

```python
before.write("어느 날, ", length=60)
```

## 실습 2-4 사전학습 도중의 모델

**먼저 예상해 보세요.** 250번, 1,000번, 3,000번 학습한 모델은 각각 어떤 글을 쓸까요? 250번만 학습한 모델도 한글 낱말을 쓸 수 있을까요?

```python
load_story_llm("step250").write("어느 날, ", length=80)
load_story_llm("step1000").write("어느 날, ", length=80)
load_story_llm("step3000").write("어느 날, ", length=80)
```

## 실습 2-5 사전학습 모델에 요청하기

**먼저 예상해 보세요.** 사전학습만 한 모델에게 "토끼가 나오는 짧은 이야기를 써 줘"라고 부탁하면, 요청대로 토끼 이야기를 새로 시작할까요?

```python
pre.ask("토끼가 나오는 짧은 이야기를 써 줘", length=150)
```

## 실습 2-6 파인튜닝 모델에 같은 요청하기

```python
sft.ask("토끼가 나오는 짧은 이야기를 써 줘", length=150)
```

## 실습 2-7 이야기가 아닌 질문하기

**먼저 예상해 보세요.** 파인튜닝 모델에게 "대한민국의 수도는 어디야?"라고 물으면 "서울"이라고 답할까요?

```python
sft.ask("대한민국의 수도는 어디야?", length=150)
```

## 실습 2-8 피드백 모델에 같은 질문하기

**먼저 예상해 보세요.** 사람의 피드백까지 학습한 모델에게 같은 질문을 하면, 이번에도 이야기를 지어낼까요?

```python
fb.ask("대한민국의 수도는 어디야?", length=150)
```

## 실습 2-9 피드백 모델에 이야기 부탁하기

```python
fb.ask("토끼가 나오는 짧은 이야기를 써 줘", length=150)
```
